```python
### Results from interFlow
agglomerator = da.data_agglomerator("FrOnRefl.parameter","SumFrefl.dat", 
                                     "./Group1-FrOnRefl_00000_templateCase")
agglomerator.show_failed_variations()
agglomerator.write_agglomerated_study_data("Group1-FrOnRefl.csv")
Group1_FrOnRefl = agglomerator.study_dataframe()
Group1_FrOnRefl = Group1_FrOnRefl.sort_index()

Dfacs = Group1_FrOnRefl.index.get_level_values("dFac").unique()

rcParams["figure.dpi"] = 400
forces=[]
for Dfac in Dfacs:
    df_subset = Group1_FrOnRefl.loc[Dfac]
    forces.append(abs(df_subset['force_y'].iloc[7000])*1000)
plt.scatter(Dfacs,forces)
#plt.vlines(1.25,0,10)
#plt.vlines(1.75,0,10)
plt.xticks([1,1.25,1.5,1.75,2])
plt.grid()
plt.xlabel("Normalized distance")
plt.ylabel("Force on the reflector (mN)")
plt.savefig('FrOnRefl_fixedValue.png',dpi=400)
```
